## [1단계] 난장판(더미) 파일 준비하기
먼저, 청소할 가짜 쓰레기 파일들이 필요합니다. 코랩 환경 내에 임의의 파일 100개를 무작위로 생성하는 코드를 제미니에게 물어보고, 아래 셀에 붙여넣어 실행하세요.

* 💡 **프롬프트 팁**: "코랩 환경의 'downloads'라는 폴더 안에 임의의 가짜 파일 100개를 만들어주는 파이썬 코드를 짜줘. 확장자는 pdf, docx, jpg, png, txt, xlsx가 섞여 있게 해줘."

# 11주차 실습: 난장판 폴더 확장자별 자동 청소하기

이 실습 노트북은 파일 확장자를 기준으로 자동 폴더 생성 및 이동을 수행하는 파이썬 스크립트를 작성하는 공간입니다. **코드를 외워서 직접 짜지 마세요.** 아래 단계에 따라 AI(Gemini)에게 요구하고 코드를 가져와서 에러를 해결해 보세요.

```python
import os
import random

# Define the target directory and file extensions
downloads_dir = 'downloads'
extensions = ['pdf', 'docx', 'jpg', 'png', 'txt', 'xlsx']

# Create the 'downloads' directory if it doesn't exist
if not os.path.exists(downloads_dir):
    os.makedirs(downloads_dir)
    print(f"'{downloads_dir}' 폴더를 생성했습니다.")
else:
    print(f"'{downloads_dir}' 폴더가 이미 존재합니다.")

# Generate 100 dummy files
num_files = 100
for i in range(num_files):
    # Generate a random filename
    filename = ''.join(random.choices('abcdefghijklmnopqrstuvwxyz0123456789', k=10))
    # Choose a random extension
    ext = random.choice(extensions)
    # Combine to form the full file path
    file_path = os.path.join(downloads_dir, f"{filename}.{ext}")

    # Create a dummy file with some content
    try:
        with open(file_path, 'w') as f:
            f.write(f"This is a dummy file named {filename}.{ext}\n")
        # print(f"Created: {file_path}") # Uncomment to see each file created
    except Exception as e:
        print(f"Error creating file {file_path}: {e}")

print(f"{num_files}개의 더미 파일이 '{downloads_dir}' 폴더에 성공적으로 생성되었습니다.")
```

## [2단계] 확장자별 자동 분류 스크립트 작성
이제 위에서 만든 수많은 파일들을 확장자별로 새 폴더를 만들어 깔끔하게 이동시켜주는 분류 코드를 제미니에게 요청하고 아래 셀에 붙여넣어 실행하세요.

* 💡 **프롬프트 팁**: "'downloads' 폴더 안에 있는 파일들을 확장자별로 확인해서, 확장자 이름과 똑같은 새 폴더를 만들고 그 안으로 파일들을 모두 이동시켜주는 파이썬 코드를 만들어 줘. 이미 폴더가 있으면 만들지 말고 그냥 이동시켜 줘."

```python
import os
import shutil

downloads_dir = 'downloads'

# Get a list of all files in the downloads directory
files = [f for f in os.listdir(downloads_dir) if os.path.isfile(os.path.join(downloads_dir, f))]

if not files:
    print(f"Thư mục '{downloads_dir}' không chứa tệp nào để phân loại.")
else:
    print(f"Bắt đầu phân loại {len(files)} tệp trong thư mục '{downloads_dir}'.")

for file_name in files:
    # Split the file name into name and extension
    name, extension = os.path.splitext(file_name)

    # Remove the leading dot from the extension (e.g., '.pdf' -> 'pdf')
    if extension:
        extension = extension[1:]
    else:
        # If there's no extension, categorize as 'no_extension'
        extension = 'no_extension'

    # Create a new directory for the extension if it doesn't exist
    destination_dir = os.path.join(downloads_dir, extension)
    if not os.path.exists(destination_dir):
        os.makedirs(destination_dir)
        print(f"Đã tạo thư mục: '{destination_dir}'")

    # Move the file to the new directory
    source_path = os.path.join(downloads_dir, file_name)
    destination_path = os.path.join(destination_dir, file_name)

    try:
        shutil.move(source_path, destination_path)
        # print(f"Đã di chuyển: '{file_name}' đến '{destination_dir}'")
    except Exception as e:
        print(f"Lỗi khi di chuyển tệp '{file_name}': {e}")

print(f"Hoàn thành phân loại tệp trong thư mục '{downloads_dir}'.")
```

```python
import os

downloads_dir = 'downloads'

print(f"Kiểm tra nội dung của thư mục '{downloads_dir}':")
for root, dirs, files in os.walk(downloads_dir):
    level = root.replace(downloads_dir, '').count(os.sep)
    indent = ' ' * 4 * (level)
    print(f'{indent}{os.path.basename(root)}/')
    subindent = ' ' * 4 * (level + 1)
    for f in files:
        print(f'{subindent}{f}')
```
